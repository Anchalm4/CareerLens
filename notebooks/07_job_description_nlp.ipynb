```python
import pandas as pd
import re

df_jobs = pd.read_csv("../data/raw/job_descriptions.csv")

print("Dataset Shape:", df_jobs.shape)
df_jobs.head()
```

```python
skill_dictionary = [
    "Python",
    "SQL",
    "Excel",
    "Power BI",
    "Tableau",
    "Statistics",
    "Machine Learning",
    "Deep Learning",
    "Data Visualization",
    "DAX",
    "Java",
    "Programming",
    "Data Structures",
    "Algorithms",
    "Communication",
    "Presentation",
    "Business Analysis",
    "Problem Solving"
]
```

```python
def extract_skills(text):
    text = text.lower()
    
    found_skills = []
    
    for skill in skill_dictionary:
        if skill.lower() in text:
            found_skills.append(skill)
    
    return found_skills


df_jobs["extracted_skills"] = df_jobs["description"].apply(extract_skills)

df_jobs[["role", "company", "extracted_skills"]]
```

```python
from collections import Counter

skill_counter = Counter()

for skills in df_jobs["extracted_skills"]:
    skill_counter.update(skills)

market_demand = pd.DataFrame(
    skill_counter.items(),
    columns=["Skill", "Job_Count"]
)

market_demand["Demand_Percentage"] = (
    market_demand["Job_Count"] / len(df_jobs) * 100
)

market_demand = market_demand.sort_values(
    "Job_Count",
    ascending=False
).reset_index(drop=True)

display(market_demand)
```

```python
market_demand.to_csv(
    "../reports/market_skill_demand.csv",
    index=False
)

df_jobs.to_csv(
    "../data/processed/job_skill_analysis.csv",
    index=False
)

print("Job description analysis completed!")
```

```python
import pandas as pd

students = pd.read_csv("../data/processed/modeling_dataset.csv")
market_demand = pd.read_csv("../reports/market_skill_demand.csv")

print("Students:", students.shape)
print("Market Skills:", market_demand.shape)

display(market_demand)
```

```python
skill_columns = {
    "Python": "python_score",
    "SQL": "sql_score",
    "Excel": "excel_score",
    "Power BI": "powerbi_score",
    "Statistics": "statistics_score",
    "Machine Learning": "ml_score",
    "Data Visualization": "data_visualization_score",
    "Java": "java_score",
    "Communication": "communication_score",
    "Presentation": "presentation_score",
    "Programming": "coding_score",
    "Problem Solving": "logical_reasoning_score"
}
```

```python
def calculate_market_skill_gap(student):

    results = []

    for skill, column in skill_columns.items():

        student_score = student[column]

        market_row = market_demand[
            market_demand["Skill"] == skill
        ]

        if not market_row.empty:
            demand = market_row.iloc[0]["Demand_Percentage"]
        else:
            demand = 0

        gap = max(0, 100 - student_score)

        priority = gap * (demand / 100)

        results.append({
            "Skill": skill,
            "Student_Score": round(student_score, 2),
            "Market_Demand": round(demand, 2),
            "Skill_Gap": round(gap, 2),
            "Priority": round(priority, 2)
        })

    return pd.DataFrame(results)
```

```python
student = students.iloc[0]

market_gap = calculate_market_skill_gap(student)

market_gap = market_gap.sort_values(
    "Priority",
    ascending=False
).reset_index(drop=True)

display(market_gap)
```

```python
top_skill_gaps = market_gap.head(5)

print("Top Skills to Improve:")
display(top_skill_gaps)
```

```python
market_gap.to_csv(
    "../reports/student_market_skill_gap.csv",
    index=False
)

print("Student market skill gap analysis completed!")
```
