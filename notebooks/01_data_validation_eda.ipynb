```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
data_path = Path("../data/raw/career_dataset.csv")

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage.head(10)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
print(
    "Duplicate student IDs:",
    df["student_id"].duplicated().sum()
)
```

```python
df.describe()
```

```python
placement_counts = df["placed"].value_counts()

print(placement_counts)
```

```python
placement_percentage = (
    df["placed"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(placement_percentage)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="placed"
)

plt.title("Placement Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")

plt.show()
```

```python
branch_placement = (
    df.groupby("branch")["placed"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print(branch_placement)
```

```python
plt.figure(figsize=(9, 5))

branch_placement.plot(
    kind="bar"
)

plt.title("Placement Rate by Engineering Branch")
plt.xlabel("Branch")
plt.ylabel("Placement Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="placed",
    y="cgpa"
)

plt.title("CGPA Distribution by Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("CGPA")

plt.show()
```

```python
internship_placement = (
    df.groupby("internship_count")["placed"]
    .mean()
    .mul(100)
)

print(internship_placement)
```

```python
plt.figure(figsize=(8, 5))

internship_placement.plot(
    kind="bar"
)

plt.title("Placement Rate by Number of Internships")
plt.xlabel("Number of Internships")
plt.ylabel("Placement Rate (%)")

plt.tight_layout()
plt.show()
```

```python
skill_columns = [
    "python_score",
    "sql_score",
    "excel_score",
    "powerbi_score",
    "statistics_score",
    "ml_score",
    "coding_score"
]

skill_placement = df.groupby("placed")[skill_columns].mean()

print(skill_placement.round(2))
```

```python
skill_placement.T.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Average Skill Scores by Placement Status")
plt.xlabel("Skill")
plt.ylabel("Average Score")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(
    include=np.number
)

correlation = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
placement_corr = (
    correlation["placed"]
    .sort_values(
        ascending=False
    )
)

print(placement_corr)
```

```python
print(
    placement_corr.head(15)
)
```

```python
placed_students = df[df["placed"] == 1]

print(
    placed_students["package_lpa"].describe()
)
```

```python
placed_students = df[df["placed"] == 1]

print(
    placed_students["package_lpa"].describe()
)
```

```python
print(
    "Median Package:",
    round(
        placed_students["package_lpa"].median(),
        2
    ),
    "LPA"
)
```

```python
salary_by_role = (
    placed_students
    .groupby("placement_role")["package_lpa"]
    .agg(["count", "mean", "median"])
    .sort_values("mean", ascending=False)
)

print(
    salary_by_role.round(2)
)
```
